```python
#Renaming Columns
df=df.rename(columns={'Date':'Sales Date'})
df.head()
```

```python
## Change datatypes
df['Sales_new']=df['Sales'].fillna(df['Sales'].mean()).astype(int)
df.head()
```

```python
df['New Value']=df['Sales'].apply(lambda x:x*2)
df.head()
```

```python
## Data Aggregating AND Grouping

df.head()
```

```python
grouped_mean=df.groupby('Product')['Sales'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['Product','Region'])['Sales'].sum()
print(grouped_sum)
```

```python
df.groupby(['Product','Region'])['Sales'].mean()
```

```python
## aggreate multiple functions
grouped_agg=df.groupby('Region')['Sales'].agg(['mean','sum','count'])

grouped_agg
```

```python
## Merging and  Joining Dataframnes
# Create sample DataFrames

df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
# Merge Dataframe on the 'Key Columns'
pd.merge(df1,df2,on="Key",how='inner')
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
